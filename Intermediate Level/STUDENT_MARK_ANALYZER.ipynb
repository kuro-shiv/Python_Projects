```python
#STUDENT MARKS ANALYZER.

import numpy as np
marks = np.array([
    [78, 85, 90],   # Student 1
    [92, 88, 79],   # Student 2
    [65, 70, 72],   # Student 3
    [89, 94, 91],   # Student 4
    [75, 80, 83]    # Student 5

])
```

```python
total_marks = np.sum(marks, axis=1)
print("Total marks per student:", total_marks)
```

```python
average_subject_marks = np.mean(marks, axis=0)
print("Average marks per subject:", average_subject_marks)
```

```python
highest_scorer_index = np.argmax(total_marks)
print("Index of highest scorer:", highest_scorer_index)
```

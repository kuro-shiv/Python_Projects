```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Roll": range(101, 136),
    "Name": [
        "Aarav", "Vivaan", "Aditya", "Vihaan", "Arjun", "Sai", "Reyansh", "Ayaan", "Krishna", "Ishaan",
        "Anaya", "Aadhya", "Myra", "Siya", "Avni", "Diya", "Meera", "Ira", "Saanvi", "Pari",
        "Yash", "Rohan", "Laksh", "Kunal", "Kabir", "Aryan", "Riya", "Tanya", "Shreya", "Naina",
        "Rahul", "Sneha", "Priya", "Dev", "Manav"
    ],
    "Branch": np.random.choice(["CS", "DS", "IT"], 35),
    "Department": np.random.choice(["Engineering", "Science"], 35),
    "Year": np.random.choice([2022, 2023, 2024], 35),
    "City": np.random.choice(["Lucknow", "Delhi", "Mumbai"], 35),
    "PPS": np.random.randint(40, 100, 35),
    "DSA": np.random.randint(20, 100, 35),
    "DS": np.random.randint(0, 100, 35),
    "COA": np.random.randint(50, 100, 35),
    "Chemistry": np.random.randint(30, 100, 35),
    "Maths": np.random.randint(10, 100, 35),
}

df = pd.DataFrame(data)
print("Initial DataFrame:")
print(df.head())
```

```python
df["TotalMarks"] = df[["Maths", "PPS", "Chemistry", "DS" , "COA", "DSA" ]].sum(axis=1)
print(df)
```

```python
cs_ds_students = df[df["Branch"].isin(["CS", "DS"])]
print("\nStudents from CS and DS:")
print(cs_ds_students)
print("-----------------------------------------------------------------------")
print(f"\nTotal CS and DS students: {cs_ds_students.shape[0]}")
```

```python
yearwise_students = df[df["Year"].isin([2022, 2023, 2024])]
print("\nYear-wise Students:")
print(yearwise_students)
```

```python
cs_2022 = df[(df["Year"] == 2022) & (df["Branch"] == "CS")]
cs_2022_count = cs_2022.shape[0]
print("\n2022 CS Students:")
print(cs_2022)
print("-----------------------------------------------------------------------")
print(f"Size of the students: {cs_2022_count}")
```

```python
ds_2023 = df[(df["Year"] == 2023) & (df["Branch"] == "DS")]
print("\n2023 DS Students:")
print(ds_2023)
print("-----------------------------------------------------------------------")
ds_2023_count = ds_2023.shape[0]
print(f"Size of the student: {ds_2023_count}")
```

```python
lucknow_students = df[df["City"] == "Lucknow"]
lucknow_student_count = lucknow_students.shape[0]
print("\nStudents from Lucknow:")
print(lucknow_students)
print("-----------------------------------------------------------------------")
print(f"Total Number of studets in lucknow: {lucknow_student_count}")
```

```python
df_dropped = df.drop(columns=["Chemistry"])
print("\nDataFrame after dropping 'Chemistry':")
df["TotalMarks"] = df[["Maths", "PPS",  "DS", "COA", "DSA" ]].sum(axis=1)
print(df_dropped.head())
```

```python
topper_by_year = df.loc[df.groupby("Year")["TotalMarks"].idxmax()]
lowscorer_by_year = df.loc[df.groupby("Year")["TotalMarks"].idxmin()]
print(f"Student topped each year: {topper_by_year}")
print(f"Student with low each year: {lowscorer_by_year}")
```

```python
low_scorers = df_dropped[df_dropped["TotalMarks"] < 300]
print(low_scorers)
```

```python
new_df = df[~df.index.isin(low_scorers.index)]
print(new_df)
print("---------------------------------------------------------")
print(f"No of students passed the exam: {new_df.shape[0]}")
```

```python
df.to_csv("Updated_Student.csv", index=False)
print("\n DataFrame saved as 'Updated_Student.csv'")
```

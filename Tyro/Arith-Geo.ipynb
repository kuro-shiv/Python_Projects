```python
def ArithGeo(arr):
    arith,geo = True, True
    for num in range(1, len(arr) - 1):
        if arr[num + 1] - arr[num] != arr[num] - arr[num-1]:
            arith = False
        if arr[num +1] / arr[num] != arr[num] / arr[num-1]:
            geo = False
    return "Arithmetic" if arith else "Geometric" if geo else -1
```

```python
arr = [2, 4, 6, 8]
ArithGeo(arr)
```

```python
def ArrayAdditionI(arr):
  arr.sort()
  largest = arr.pop()
  result = [False]
  def findCombo(sum=0,i=0):
    if sum is largest:
      result[0] = True
      return
    while i < len(arr):
      sum += arr[i]
      findCombo(sum,i+1)
      sum -= arr[i]
      i += 1
  findCombo()
  return result[0]
```

```python
arr = [4, 6, 23, 10, 1, 3]
ArrayAdditionI(arr)
```

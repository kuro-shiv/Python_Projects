```python
def CountingMinutesI(str):
    times = str.split("-")
    time1 = times[0][:-2], times[0][-2:]
    time2 = times[1][:-2], times[1][-2:]
    hours12 = 720
    hours24 = 1440

    def getMinutes(time, amOrPm):
        arr = time.split(":")
        hours = arr[0]
        minutes = arr[1]
        res = 0
        if amOrPm == "pm":
            res = int(hours) * 60 + hours12 + int(minutes)
        if amOrPm == "am":
            res = int(hours) * 60 + int(minutes)
        return res

    min1 = getMinutes(*time1)
    min2 = getMinutes(*time2)

    if min2 > min1:
        return min2 - min1
    elif min1 > min2:
        return hours24 - min1 + min2
```

```python
str = "9:00am-10:00am"
CountingMinutesI(str)
```

```python
str = "9:00am-10:00pm"
CountingMinutesI(str)
```

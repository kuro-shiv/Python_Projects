```python
from functiontools import reduce
def MeanMode(arr):
  mean = reduce(lambda x,y: x + y, arr)/float(len(arr))
  def getMode(arr):
      storage = {}
      for num in arr:
        try:
            storage[num] = storage[num] + 1
        except: storage[num] = 1
      tempMode = max(storage, key=lambda k: storage[k])
      return tempMode   
  mode = getMode(arr)
  return 1 if mean == mode else 0 
```

```python
arr = [5, 3, 3, 3, 1]
MeanMode(arr)
```

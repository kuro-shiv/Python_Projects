```python
def SecondGreatLow(arr): 
  arr = sorted(arr)

  for i in range(0, len(arr)-1):
        if arr[i] == arr[i+1]:
            low = i+1
        elif arr[i] < arr[i+1]:
            low = i+1
            break          
  return str(arr[low]) + ' ' + str(arr[-2]) 
```

```python
arr = [7, 7, 12, 98, 106]
SecondGreatLow(arr)
```
